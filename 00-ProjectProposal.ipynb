## Research Question

Can machine learning models predict whether betting favorites will cover the spread more accurately than random chance by learning from historical quarter-by-quarter scoring patterns from the 2007-2024 NBA games? Specifically, we want to investigate whether scoring distribution across quarters contains predictive information that improves upon normal betting strategies.

## Background and Prior Work

The sports betting market represents a unique testing ground for market efficiency theories. The Efficient Market Hypothesis, originally developed for financial markets, suggests that prices fully reflect all the available information. When applied specifically to the context of sports betting, this suggests that betting odds should be an accurate representation of the true probability of game outcomes. This should, in theory, make consistent profitability nearly impossible simply because their models are designed to be as accurate about the predictions as possible.

The application of machine learning towards NBA game prediction has been a hugely popular endeavor undertaken by many in recent reviews. However, most papers and betting lines don't ever seem to take into account game line statistics like quarter-by-quarter scoring. Our research aims to see if this choice, if deliberate, was a good one or not by evaluating whether these statistics do have some meaningful impact on the end result of the game.

## Hypothesis


We hypothesize that machine learning models trained from historical quarter-by-quarter scoring patterns from the 2007-2024 NBA games will achieve a higher prediction accuracy than random chance for whether betting favorites will cover the spread. We predict that the scoring distribution across quarters contains predictive information that improves upon normal betting strategies.

## Data

1. The ideal dataset would have the following characteristics:
   1. Extensive Variable Options
      A. The ideal dataset would include variables that cover the normal game data such as date, season, regular season vs. playoffs. It would also have variables that differentiate between home team and away team. There should also be variables such as points scored in each quarter by each team, final scores for home teams vs away teams, prediction target (whether the popular betting choice covered the spread), and also prediction data such as the point spread. Quarter-by-quarter scores are important since it illustrates the point distribution throughout an entire game. It is indicative of more than just the final score, but whether a team was down and came back up to close the gap, the game had a close score throughout, or if a team lost their lead, etc.
   2. Large Quantity of Observations
      A. It is ideal to have a large sample size, around 10,000-20,000 games. This allows us to reduce the chance of overfitting. Additionally, with a large data set, data can be split up into subsets so that the model can learn to identify patterns from one portion of the dataset, adjust and improve the model from another portion of the dataset, and test the model on the remaining portion to evaluate how it works on games it has never seen. Overall, this allows for the reliability of the model’s predictions to be improved while maintaining statistical significance.
   3. Objective Measures for Data Collection
      A. Game scores, final and quarterly, are collected from official NBA game records. Point spreads and favorite indicators are to be collected from sports betting market data. Sports analytics platforms put together datasets with this information, allowing the two to be analyzed simultaneously.
   4. Easy-to-Read Data Storage
      A.  Data should be stored in tabular format such as CSV files. Each row represents a single game while each column represents a variable of that game such as quarter-by-quarter scores, point spread, etc.
2. We found an interesting dataset that contained the specific information we need to answer our question.
   1. The data is located on Kaggle at the following link: https://www.kaggle.com/datasets/cviaxmiwnptr/nba-betting-data-october-2007-to-june-2024 
   2. There are 27 available variables for us to use in our research. There are important game identifier variables like season, date, and regular season game or playoff game indicators. There are also quarter-by-quarter scoring variables that help us answer our question specifically in regards to whether quarter-by-quarter scoring has some sort of meaningful impact on the outcome of a game that normal betting odds don’t take into account. Finally, we also have betting market data regarding the spread, total, and who’s favored so we have a benchmark to compare our own predictions to.
